```python
#convert numiric column
df['transaction_qty'] = pd.to_numeric(
    df['transaction_qty'], errors='coerce'
)

df['unit_price'] = pd.to_numeric(
    df['unit_price'], errors='coerce'
)
```

```python
# validate quantity and price
print("Invalid quantities:",
      (df['transaction_qty'] <= 0).sum())

print("Invalid prices:",
      (df['unit_price'] <= 0).sum())
```

```python
# check product category
print(df['product_category'].value_counts())

#remove Space
print(df['product_category'].unique())
```

```python
#remove space and text value
text_columns = [
    'store_location',
    'product_category',
    'product_type',
    'product_detail'
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
# Check product identifiers and required values
print("Missing product IDs:", df['product_id'].isnull().sum())
print("Missing prices:", df['unit_price'].isnull().sum())
print("Missing quantities:", df['transaction_qty'].isnull().sum())
```

```python
# calculate Revenue
df['Revenue'] = (
    df['transaction_qty'] * df['unit_price']
)

print(df[['transaction_qty', 'unit_price', 'Revenue']].head())
```

```python
# Aggregate revenue  by Product
product_revenue = (
    df.groupby(['product_id', 'product_detail'])
      .agg(
          Total_Units=('transaction_qty', 'sum'),
          Total_Revenue=('Revenue', 'sum')
      )
      .reset_index()
)

print(product_revenue.head())
```

```python
# Aggregate revenue by product type
type_revenue = (
    df.groupby('product_type')['Revenue']
      .sum()
      .reset_index(name='Total_Revenue')
      .sort_values('Total_Revenue', ascending=False)
)
type_revenue.head()
```

```python
# Aggregate revenue by product category
category_revenue = (
    df.groupby('product_category')['Revenue']
      .sum()
      .reset_index(name='Total_Revenue')
      .sort_values('Total_Revenue', ascending=False)
)

print(category_revenue)
```

```python
# Product Popularity Analysis
# Total units sold per product

product_popularity = (
    df.groupby(['product_id', 'product_detail'])['transaction_qty']
      .sum()
      .reset_index(name='Total_Units_Sold')
      .sort_values('Total_Units_Sold', ascending=False)
)

product_popularity.head(10)
```

```python
#  Ranking products by sales volume

product_popularity['Volume_Rank'] = (
    product_popularity['Total_Units_Sold']
    .rank(method='dense', ascending=False)
    .astype(int)
)

product_popularity.head(10)
```

```python
# identify top and bottom performers

print("Top 10 popular products:")
print(product_popularity.head(10))

print("Bottom 10 products by units sold:")
print(product_popularity.sort_values(
    'Total_Units_Sold'
).head(10))
```

```python
# Revenue Contribution Analysis
#Total revenue per product

product_revenue['Revenue_Share_Percent'] = (
    product_revenue['Total_Revenue']
    / product_revenue['Total_Revenue'].sum()
    * 100
).round(2)

product_revenue['Revenue_Rank'] = (
    product_revenue['Total_Revenue']
    .rank(method='dense', ascending=False)
    .astype(int)
)

product_revenue = product_revenue.sort_values(
    'Total_Revenue', ascending=False
)

print(product_revenue.head())
```

```python
 #compare volume rank and revenue rank
comparison = product_popularity.merge(
    product_revenue[
        ['product_id', 'Total_Revenue',
         'Revenue_Share_Percent', 'Revenue_Rank']
    ],
    on='product_id',
    how='left'
)

print(comparison[
    ['product_id', 'product_detail', 'Total_Units_Sold',
     'Volume_Rank', 'Total_Revenue', 'Revenue_Rank']
].head())
```

```python
#Category & Product-Type Performance
# Calculate revenue share by category (Coffee, Tea, Chocolate)

category_revenue['Revenue_Share_Percent'] = (
    category_revenue['Total_Revenue']
    / category_revenue['Total_Revenue'].sum()
    * 100
).round(2)

category_revenue
```

```python
# calculate Product-type contribution within each category

type_category = (
    df.groupby(['product_category', 'product_type'])['Revenue']
      .sum()
      .reset_index(name='Type_Revenue')
)

type_category['Category_Total'] = (
    type_category.groupby('product_category')['Type_Revenue']
                 .transform('sum')
)

type_category['Type_Contribution_Percent'] = (
    type_category['Type_Revenue']
    / type_category['Category_Total'] * 100
).round(2)

type_category.sort_values(
    ['product_category', 'Type_Revenue'],
    ascending=[True, False]
).head(10)
```

```python
# Dependence analysis on core categories

category_share = category_revenue.sort_values(
    'Revenue_Share_Percent', ascending=False
)

top_share = category_share.head()['Revenue_Share_Percent'].sum()

print("Revenue share of top  categories:", round(top_share, 2), "%")
```

```python
# Revenue Concentration & Menu Balance
#Pareto (80/20) analysis

pareto = product_revenue.sort_values(
    'Total_Revenue', ascending=False
).copy()

pareto['Cumulative_Revenue'] = pareto['Total_Revenue'].cumsum()

pareto['Cumulative_Revenue_Percent'] = (
    pareto['Cumulative_Revenue']
    / pareto['Total_Revenue'].sum() * 100
)

pareto['Product_Number'] = range(1, len(pareto) + 1)

pareto[
    ['product_id', 'product_detail', 'Total_Revenue',
     'Cumulative_Revenue_Percent']
].head(10)
```

```python
# highest performing product
products_for_80 = (
    pareto['Cumulative_Revenue_Percent'] < 80
).sum() + 1

print("Products needed to reach approximately 80% revenue:",
      products_for_80)
```

```python
# Identify Revenue anchors (few products driving most revenue)
revenue_anchors = pareto[
    pareto['Cumulative_Revenue_Percent'] <= 80
]

print("Revenue anchor products:")
print(revenue_anchors[
    ['product_id', 'product_detail',
     'Total_Revenue', 'Revenue_Share_Percent']
].head(10))
```

```python
#Identify Long-tail products
long_tail = pareto[
    pareto['Revenue_Share_Percent'] < 1
]

print("Long-tail products:")
print(long_tail[
    ['product_id', 'product_detail',
     'Total_Revenue', 'Revenue_Share_Percent']
].head(10))
```

```python
# Evaluate menu diversification risk

with pd.ExcelWriter(
    '/content/coffee_roaster_analysis.xlsx',
    engine='openpyxl'
) as writer:

    df.to_excel(writer, sheet_name='Cleaned_Data', index=False)
    product_popularity.to_excel(
        writer, sheet_name='Popularity', index=False
    )
    product_revenue.to_excel(
        writer, sheet_name='Product_Revenue', index=False
    )
    category_revenue.to_excel(
        writer, sheet_name='Category_Revenue', index=False
    )
    type_category.to_excel(
        writer, sheet_name='Type_Performance', index=False
    )
    pareto.to_excel(
        writer, sheet_name='Pareto_Analysis', index=False
    )

    print("Analysis workbook created successfully!")
```
