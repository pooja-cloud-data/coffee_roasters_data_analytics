```python
import pandas as pd
filename = list(uploaded.keys())[0]

df = pd.read_excel("Afficionado Coffee Roasters.xlsx")

#print(df)

# display first five row
print(df.head())

# check columns name
print(df.columns.tolist())

#show rows & column count
print(df.shape)

#show columns and missing value
print(df.info())

```

```python
# check missing value
#print(df.isnull())

#check missing value and count
#print(df.isnull().sum())

#check duplicate rows
print(df.duplicated())

#check duplicated rows and count
print(df.duplicated().sum())

#check data type
print(df.dtypes)
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Clean product identifiers
df["product_id"] = df["product_id"].astype("string").str.strip()

##convert price to numeric
df["unit_price"] = pd.to_numeric(
    df["unit_price"],
    errors="coerce"
)

#check missing producet id
print(df["product_id"].isna().sum())

#check missing prices
print(df["unit_price"].isna().sum())

#check total record
print("Total records:", len(df))

#check missing product id
print(
    "Missing product IDs:",
    df["product_id"].isna().sum()
)

#check unit price invalid
print(
    "Invalid prices:",
    (df["unit_price"].isna() |
     (df["unit_price"] <= 0)).sum()
)

# check transaction quantity invalid
print(
    "Invalid quantities:",
    (df["transaction_qty"].isna() |
     (df["transaction_qty"] <= 0)).sum()
)

```

```python
#  product catalogue
valid_product_ids = {"P101", "P102", "P103"}

df["Unknown_product"] = ~df["product_id"].isin(
    valid_product_ids
)

print(df[["product_id", "Unknown_product"]])
```

```python
 #remove duplicate rows
df = df.drop_duplicates()

print("shape after removing duplicated:", df.shape)

# Remove unneccessary sapce from column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print(df.columns.tolist())

```

```python
print("Total transactions:", len(df))
print("Toatal products:", len(df))
```

```python
#revenue computation
#column name quantity price
df["Revenue"] = df["transaction_qty"] * df["unit_price"]

display(df.head())
```

```python
df["Revenue"] = (
    df["transaction_qty"] * df["unit_price"])


df["transaction_time"] = pd.to_datetime(
    df["transaction_time"].astype(str),
    errors="coerce")


df["hour"] = df["transaction_time"].dt.hour

product_revenue = df.groupby('product_category')['Revenue'].sum() # check product name

print(product_revenue.sort_values(ascending=False))
```

```python
#calculate revenue
print(df.groupby("product_type")["Revenue"].sum())
print(df.groupby("product_category")["Revenue"].sum())
print(df.groupby("transaction_qty")["Revenue"].sum())
print(df.groupby("unit_price")["Revenue"].sum())

```

```python
#calculate total sold products
product_sales = (
    df.groupby("product_category")["transaction_qty"]
      .sum()
      .reset_index()
)

product_sales.columns = ["Product", "Total_Units_Sold"]

#print(product_sales)
product_sales.head(10)
```

```python
#total units sold per product
product_sales = (
    df.groupby("product_type")["transaction_id"]
    .sum()
      .reset_index()
)

product_sales.columns = ["Product", "Total_Units_Sold"]

#print(product_sales)
product_sales.head(30)
```

```python
# Rank product sales by volume
product_sales = product_sales.sort_values(
    by="Total_Units_Sold",
    ascending=False
)

product_sales["Sales_Rank"] = (
    product_sales["Total_Units_Sold"]
    .rank(method="min", ascending=False)
    .astype(int)
)

#print(product_sales)
product_sales.head(10)
```

```python
#Identify top and bottom performers

top_product = product_sales.loc[
    product_sales["Total_Units_Sold"].idxmax()
]

bottom_product = product_sales.loc[
    product_sales["Total_Units_Sold"].idxmin()
]

print("Top-performing product:", top_product["Product"])
print("Units sold:", top_product["Total_Units_Sold"])

print("\nBottom-performing product:", bottom_product["Product"])
print("Units sold:", bottom_product["Total_Units_Sold"])
```

```python
# calculate total revenue per product
product_revenue = (
    df.groupby("product_category")["Revenue"]
      .sum()
      .reset_index()
)

product_revenue.columns = ["Product", "Total_Revenue"]

#print(product_revenue)
product_revenue.head(10)
```

```python
#Revenue share (%) of each product
total_revenue = product_revenue["Total_Revenue"].sum()

product_revenue["Revenue_Share_Percent"] = (
    product_revenue["Total_Revenue"] / total_revenue
) * 100

product_revenue["Revenue_Share_Percent"] = (
    product_revenue["Revenue_Share_Percent"].round(2)
)

product_revenue.head(10)
```

```python
#Comparison between volume rank and revenue rank

# Calculate total units sold per product
product_volume = (
    df.groupby("product_type")["transaction_qty"]
      .sum()
      .reset_index()
)

product_volume.columns = ["Product", "Total_Units_Sold"]

# Combine volume and revenue data
comparison = product_volume.merge(
    product_revenue,
    on="Product"
)

# Rank by units sold
comparison["Volume_Rank"] = (
    comparison["Total_Units_Sold"]
    .rank(method="min", ascending=False)
    .astype(int)
)

comparison["Volume_Rank"].head()
# Rank by revenue
comparison["Revenue_Rank"] = (
    comparison["Total_Revenue"]
    .rank(method="min", ascending=False)
    .astype(int)
)

# Sort by volume rank
comparison = comparison.sort_values("Volume_Rank")

comparison.head()
```

```python
#Revenue share by category (Coffee, Tea, Chocolate)
# Total revenue for each category
category_revenue = (
    df.groupby("product_category")["Revenue"]
      .sum()
      .reset_index()
)

category_revenue.columns = [
    "Product_Category", "Total_Revenue"
]

# Calculate percentage contribution
total_revenue = df["Revenue"].sum()

category_revenue["Revenue_Share_Percent"] = (
    category_revenue["Total_Revenue"] / total_revenue * 100
).round(2)

# Show highest revenue category first
category_revenue = category_revenue.sort_values(
    "Total_Revenue", ascending=False
)

category_revenue.head(10)
```

```python
#Product-type contribution within each category
# Revenue by category and product type
type_revenue = (
    df.groupby(["product_category", "product_type"])["Revenue"]
      .sum()
      .reset_index()
)

type_revenue.columns = [
    "Product_Category",
    "Product_Type",
    "Type_Revenue"
]

# Total revenue within each category
type_revenue["Category_Total"] = (
    type_revenue.groupby("Product_Category")["Type_Revenue"]
    .transform("sum")
)
# Percentage contribution within the category
type_revenue["Type_Contribution_Percent"] = (
    type_revenue["Type_Revenue"] /
    type_revenue["Category_Total"] * 100
).round(2)

type_revenue.head(30)
```
